# 📘 Backend Software Engineering Notes

## 1. What is API Development

**API development is the process of designing and building interfaces that allow different software applications or components to communicate and exchange data with each other.**

### সহজ বাংলা

API Development মানে এমন একটা system/interface বানানো যার মাধ্যমে **client এবং server request ও response-এর মাধ্যমে data exchange করতে পারে।**

### Easy to Remember

**Client → Request → API → Server → Response**

---

## 2. GET vs POST

### GET

**GET is used to retrieve data from a server.**

### সহজ বাংলা

Server থেকে data **নিয়ে আসা/read করার** জন্য GET ব্যবহার করি।

Example:

`GET /api/products`

### POST

**POST is used to send data to a server, usually to create a new resource.**

### সহজ বাংলা

Server-এ নতুন data **পাঠানো/create করার** জন্য POST ব্যবহার করি।

### Easy to Remember

**GET → Retrieve Data**  
**POST → Create/Send Data**

---

## 3. POST Route

### Interview Definition

**A POST route is used to handle POST requests, typically to create a new resource on the server.**

### সহজ বাংলা

Client যখন server-এ নতুন data পাঠিয়ে **কিছু create করতে চায়**, তখন POST route সেই request handle করে।

Example:

`POST /api/products`

**Client → Product Data → POST Route → New Product Created**

---

## 4. Data Modeling

### Interview Definition

**Data modeling is the process of defining how data is structured, organized, and related within an application or database.**

### সহজ বাংলা

Application-এর data **কীভাবে structure এবং organize করা হবে** সেটা design করাই Data Modeling।

Example:

```json
{
  "id": 1,
  "name": "Laptop",
  "price": 1000,
  "category": "Electronics",
  "rating": 4.5
}
```

### Easy to Remember

**Data Modeling = How our data is structured**

---

## 5. Router

### Interview Definition

**A router maps incoming HTTP requests to the appropriate controller or route handler based on the request method and URL.**

### সহজ বাংলা

Router decide করে **কোন URL + HTTP method-এর request কোন controller/function-এর কাছে যাবে।**

Example:

`GET /products` → `getProducts()`

`GET /products/5` → `getProductById()`

### Easy to Remember

**Router = Where should the request go?**

---

## 6. Middleware

### Interview Definition

**Middleware is a function that runs during the request-response cycle and can inspect, modify, reject, or pass a request to the next step.**

### সহজ বাংলা

Middleware request-এর **মাঝখানে দাঁড়িয়ে check/process করে**।

Examples:

- Token Check
- User-Agent Check
- Audit Logger
- Rate Limiter

সব ঠিক থাকলে:

`next()`

দিয়ে request সামনে পাঠায়।

### Easy to Remember

**Request → Middleware → Router → Controller**

---

## 7. Intercept

### Interview Definition

**Intercepting a request means capturing it before it reaches its final destination so it can be checked, modified, logged, or rejected.**

### সহজ বাংলা

Intercept মানে request controller-এর কাছে যাওয়ার **আগে মাঝখানে ধরে check/process করা।**

Example:

**Request → Intercept → Check → Valid?**

Valid → `next()` ✅  
Invalid → Reject ❌

### Easy to Remember

**Intercept = Catch the request before it reaches its destination**

---

## 8. Controller

### Interview Definition

**A controller handles incoming requests, coordinates the required application logic or services, and sends the appropriate response back to the client.**

### সহজ বাংলা

Controller request পাওয়ার পর প্রয়োজনীয় **logic/service call করে এবং client-কে response দেয়।**

Example:

```javascript
const getProducts = (req, res) => {
    res.status(200).json(products);
};
```

### Easy to Remember

**Controller = Handle Request → Logic → Response**

---

## 9. Business Layer

### Interview Definition

**The business layer contains the core business rules and application logic that determine how the system should behave.**

### সহজ বাংলা

Application-এর **main rules এবং actual logic** যেখানে থাকে সেটাই Business Layer।

Examples:

- Discount calculate করা
- Order valid কিনা check করা
- Total price calculate করা
- User eligible কিনা check করা

### Easy to Remember

**Business Layer = Core Application Rules**

---

# ⭐ Backend Request Flow

**Client Request**  
↓  
**Middleware** → Check / Log / Validate  
↓  
**Router** → কোথায় যাবে?  
↓  
**Controller** → Request handle করে  
↓  
**Business Layer** → Main logic/rules  
↓  
**JSON / Database** → Data  
↓  
**Response** → Client

### Short Version

**Request → Middleware → Router → Controller → Business Logic → Data → Response**

# 📘 Logging & API Security


## 1. What is Logging?

**Logging is the process of recording important events and activities that happen in an application.**

### সহজ বাংলা

Application/server-এ **কী কী হচ্ছে তার information record করে রাখা** হলো Logging।

### Easy to Remember

**Logging = Record Application Activities**

---

## 2. What is an Audit Logger?

### Interview Definition

**An audit logger records important user or API activities so they can be reviewed later.**

### সহজ বাংলা

API/server-এ **কে, কখন, কী request বা activity করেছে তার history রাখা** হলো Audit Logging।

### Example

`GET /api/products` → Method, URL, time, User-Agent save করা।

### Easy to Remember

**Audit Logger = Activity History**

---

## 3. What is Authentication?

### Interview Definition

**Authentication is the process of verifying who a user is.**

### সহজ বাংলা

User **আসলেই কে**, সেটা verify করাই Authentication।

### Example

Username + Password দিয়ে login করা।

### Easy to Remember

**Authentication = Who are you?**

---

## 4. What is Authorization?

### Interview Definition

**Authorization determines what an authenticated user is allowed to access or do.**

### সহজ বাংলা

User login করার পরে সে **কোন resource access করতে পারবে বা কী করতে পারবে**, সেটা decide করাই Authorization।

### Example

Admin → Delete product ✅  
Normal User → Delete product ❌

### Easy to Remember

**Authorization = What are you allowed to do?**

### Authentication vs Authorization

**Authentication → Who are you?**  
**Authorization → What can you do?**

---

## 5. What is a Token / API Key?

### Interview Definition

**A token or API key is a credential sent with a request to identify or authenticate a client or user.**

### সহজ বাংলা

Client request-এর সাথে একটি **special credential/key** পাঠায়, যেটা server check করে request valid কিনা বুঝতে পারে।

### Example

`/api/products?token=123`

Server checks:

`123` → Valid ✅  
Wrong token → Unauthorized ❌

### Easy to Remember

**Token/API Key = Credential for API Access**

> Real applications should not normally put sensitive tokens in the URL query string.

---

## 6. What is a User-Agent?

### Interview Definition

**A User-Agent is an HTTP header that provides information about the client software making the request.**

### সহজ বাংলা

কোন **browser, application, বা tool** থেকে request আসছে তার information User-Agent header-এ থাকতে পারে।

### Examples

- Chrome
- Mozilla/Firefox
- curl
- Postman
- python-requests

### Easy to Remember

**User-Agent = Information about the requesting client**

> User-Agent can be changed or spoofed, so it should not be used as strong authentication.

---

## 7. What is Rate Limiting?

### Interview Definition

**Rate limiting restricts how many requests a client can make within a specific period of time.**

### সহজ বাংলা

একজন client **নির্দিষ্ট সময়ের মধ্যে কতগুলো request করতে পারবে**, সেটা limit করাই Rate Limiting।

### Example

10 seconds → Maximum 5 requests

1 → ✅  
2 → ✅  
3 → ✅  
4 → ✅  
5 → ✅  
6 → ❌

### Easy to Remember

**Rate Limiting = Control Request Frequency**

Limit exceed করলে:

**429 Too Many Requests**

---

## 8. Important HTTP Status Codes

### 200 — OK

**The request was successful.**

সহজ বাংলা: Request successfully complete হয়েছে।

---

### 201 — Created

**A new resource was successfully created.**

সহজ বাংলা: নতুন data/resource successfully create হয়েছে।

Example: Successful POST request.

---

### 400 — Bad Request

**The client sent an invalid request.**

সহজ বাংলা: Client-এর request/data-তে problem আছে।

---

### 401 — Unauthorized

**Authentication is required or has failed.**

সহজ বাংলা: User/client valid authentication provide করেনি।

**Remember: 401 = Who are you?**

---

### 403 — Forbidden

**The server understands the request but does not allow access.**

সহজ বাংলা: Server request বুঝেছে, কিন্তু access করার permission নেই।

**Remember: 403 = You cannot access this.**

---

### 404 — Not Found

**The requested resource could not be found.**

সহজ বাংলা: Requested data/resource পাওয়া যায়নি।

Example:

`GET /products/6`

কিন্তু product `6` নেই → **404**

---

### 429 — Too Many Requests

**The client has exceeded the allowed request limit.**

সহজ বাংলা: খুব বেশি request পাঠানোর কারণে Rate Limiter request block করেছে।

---

### 500 — Internal Server Error

**An unexpected error occurred on the server.**

সহজ বাংলা: Server-এর ভিতরে unexpected problem হয়েছে।

---

# 📘 Object-Oriented Programming (OOP)


## 1. What is a Class?

**A class is a blueprint or template for creating objects. It defines the properties and behaviors that its objects can have.**

### সহজ বাংলা

Class হলো object বানানোর **blueprint/template**।

Example:

Class → Student

Objects:
- Sayem
- John
- Alex

### Easy to Remember

**Class = Blueprint**

---

## 2. What is an Object?

**An object is an instance of a class that contains data and can perform actions defined by the class.**

### সহজ বাংলা

**Object হলো class থেকে তৈরি হওয়া actual instance।** এর নিজের data থাকে এবং class-এর methods ব্যবহার করতে পারে।

### Easy to Remember

**Class = Blueprint**  
**Object = Actual thing created from the blueprint**

---

## 3. Properties and Methods

**Properties represent the data or state of an object, while methods define the behaviors or actions that the object can perform.**

### সহজ বাংলা

**Properties = Object-এর data**

যেমন:
- name
- age
- location

**Methods = Object কী কাজ করতে পারে**

যেমন:
- `displayInfo()`
- `calculateAge()`

---

##4. What is a Constructor?

**A constructor is a special method that is automatically called when an object is created. It is mainly used to initialize the object's data.**

### সহজ বাংলা

**Constructor হলো class-এর একটি special method, যেটা object তৈরি হওয়ার সাথে সাথে automatically call হয়।**

এর main কাজ হলো object-এর **initial data/value set করা।**

---

## 5. Encapsulation

**Encapsulation is the practice of restricting direct access to an object's internal data and providing controlled access through methods.**

### সহজ বাংলা

Encapsulation-এর main idea হলো data-তে direct access না দিয়ে method-এর মাধ্যমে controlled access দেওয়া।।

---

## 6. Inheritance

**Inheritance allows a child class to inherit properties and methods from a parent class.**

### সহজ বাংলা

একটা **child class**, parent class-এর **properties এবং methods পেয়ে/ব্যবহার করতে পারে**—এটাই Inheritance।

### Why Do We Use Inheritance?

**Main purpose → Code Reusability**

একই code child classes-এ বারবার লেখার প্রয়োজন হয় না।

---

## 7. What is Method Overriding?

**Method overriding is when a child class provides its own implementation of a method already defined in the parent class.**

### সহজ বাংলা

Parent class-এর method-কে child class নিজের প্রয়োজন অনুযায়ী **change করে ব্যবহার করে**।

### Real-World Example

```python
class Payment:
    def pay(self, amount):
        print("Processing payment")


class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Paying ${amount} using Credit Card")
```

### এখানে Overriding কোথায়?

Parent class বলেছে:

```python
def pay(self, amount):
    print("Processing payment")
```

কিন্তু `CreditCardPayment` বলছে:

> Payment তো করব, কিন্তু **Credit Card-এর নিজের process অনুযায়ী করব।**

তাই child class একই `pay()` method নিজের মতো করে define করেছে।

---

## 8. Polymorphism

**Polymorphism means the same method can behave differently for different objects.**

### সহজ বাংলা

একই **method**, different object অনুযায়ী **different behavior** দেখায়—এটাই Polymorphism।

### Real-World Example

```python
class Notification:
    def send(self):
        print("Sending Notification")

class Email(Notification):
    def send(self):
        print("Sending Email")

class SMS(Notification):
    def send(self):
        print("Sending SMS")
```

### এখানে কী হচ্ছে?

`Email` এবং `SMS` parent-এর `send()` method নিজের মতো করে change করেছে।

**এটা → Method Overriding**

তারপর:

```python
email = Email()
sms = SMS()

email.send()  # Sending Email
sms.send()    # Sending SMS
```

Same `send()` method different objects-এর জন্য different behavior করছে।

**এটা → Polymorphism**

### Easy to Remember

**Poly = Many**  
**Morph = Forms**

---

## 9. Abstraction

**Abstraction means hiding unnecessary implementation details and showing only the essential functionality to the user.**

### সহজ বাংলা

ভেতরে **কীভাবে কাজ হচ্ছে সেটা hide করে**, শুধু প্রয়োজনীয় functionality দেখানো হলো **Abstraction**।

### Real-Life Example

ATM-এ আমরা শুধু:

**Withdraw → Money**

দেখি। কিন্তু bank server, balance check, transaction processing কীভাবে হচ্ছে—সেগুলো আমাদের জানতে হয় না।

---

## 10. Method Overloading

**Method overloading means having multiple methods with the same name but different parameters.**

### সহজ বাংলা

একই **method name**, কিন্তু **parameters আলাদা** হলে তাকে Method Overloading বলে।

### Example

```java
add(int a, int b)
add(int a, int b, int c)
```

Method same → `add()`  
Parameters different → 2 vs 3 parameters

---

## 11. Why Do We Need OOP Over Procedural Programming?

### Interview Definition

**OOP is preferred for large programs because it organizes code using objects, allows data hiding, and makes code easier to reuse, manage, and maintain.**

### সহজ বাংলা

Procedural Programming-এ code মূলত **functions/procedures** দিয়ে organize করা হয়।

OOP-তে **objects** ব্যবহার করে data এবং methods একসাথে organize করা যায়। এছাড়া **data hide/protect** করা যায় এবং বড় code সহজে manage করা যায়।

### Main Advantages of OOP

- **Data Hiding** → Data protect করা যায়
- **Reusability** → একই code আবার use করা যায়
- **Modularity** → Code আলাদা class/object-এ organize করা যায়
- **Maintainability** → Code update এবং maintain করা সহজ
- **Manageability** → Large software manage করা সহজ

### Easy to Remember

**OOP = Objects + Data Hiding + Reusability + Easy Maintenance**
