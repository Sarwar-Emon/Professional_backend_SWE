# Git & GitHub Essential Commands

## 1. First-Time Setup

`git init` → Initialize Git repository

### Create `.gitignore` Before Adding Files

`touch .gitignore`

→ Creates a `.gitignore` file.

Add unnecessary files/folders inside `.gitignore`:

`node_modules/`  
`.DS_Store`  
`.env`  
`.env.*`  
`venv/`  
`.venv/`  
`__pycache__/`  
`*.pyc`  
`.vscode/`  
`*.log`

> **Why?** `.gitignore` tells Git which files/folders should NOT be tracked or uploaded to GitHub.

### Now Add and Commit

`git status` → Check files

`git add .` → Stage all files except ignored files

`git commit -m "Initial commit"` → Save a snapshot

`git remote add origin <GitHub-URL>` → Connect local repo to GitHub

`git branch -M main` → Set branch to `main`

`git push -u origin main` → First push to GitHub

---

## 2. Regular GitHub Upload

`git status` → Check changes

`git add .` → Stage changes

`git commit -m "Updated project"` → Commit changes

`git push` → Upload to GitHub

> **Remember:** `status → add → commit → push`

---

## 3. Useful Commands

`git log --oneline` → See commit history

`git diff` → See code changes

`git branch` → See branches

`git remote -v` → See connected GitHub repository

`git pull` → Get latest changes from GitHub

---

## 4. Fix Common Mistakes

`git restore --staged .` → Unstage everything

`git restore filename` → Discard uncommitted changes ⚠️

`git rm --cached filename` → Stop tracking a file

`git rm -r --cached foldername` → Stop tracking a folder

---

# 🌐 Backend Fundamentals

## 1. What is Localhost?

**Localhost** refers to **your own computer** in networking.

> **Localhost = This computer**

- IPv4 address: `127.0.0.1`
- IPv6 address: `::1`
- Requests to localhost stay on your computer and **do not go to the internet**.

### Example

If a Node.js server is running on port `3000`:

`http://localhost:3000`

means:

**Connect to the server running on my own computer at port 3000.**

### Common Uses

- **Development & Testing** → Test applications locally before deploying them.
- **Local Servers** → Run Node.js, Apache, Nginx, etc. on your computer.
- **Loopback** → Network communication stays within your own machine.

---

## 2. What is a Web Server?

A **web server** is software that **receives HTTP requests from clients and sends HTTP responses back**.

### Simple Flow

**Client/Browser → HTTP Request → Web Server → HTTP Response → Client/Browser**

Example:

1. Browser requests `/home`.
2. Web server receives the request.
3. Server processes the request.
4. Server sends HTML, JSON, images, etc. back.
5. Browser displays/uses the response.

> **Easy Definition:** A web server listens for HTTP requests and sends responses.

### Static vs Dynamic

- **Static Server** → Sends existing files such as HTML, CSS, images.
- **Dynamic Server** → Generates/processes content using backend code, databases, APIs, etc.

---

## 3. What is a Process?

A **process** is a **program that is currently running**.

For example:

You have VS Code installed → **Program**

You open VS Code → it creates one or more **processes**.

A process can have:

- **PID (Process ID)** → Unique ID assigned by the OS.
- **Memory (RAM)** → Memory used while running.
- **Threads** → Units of execution inside a process.
- **Resources** → CPU, files, network connections, etc.

> **Easy Definition:** Program = code/software; Process = that program currently running.

---

## 4. Process and Port

A server **process can listen on a port** to receive network requests.

Think of:

**IP Address = Building address**

**Port = Door number**

**Process = Person/service waiting behind that door**

Example:

`localhost:3000`

- `localhost` → My computer
- `3000` → Port
- Node.js/Express process → Listening for requests on that port

### Example

`app.listen(3000);`

This means:

**Start the server and listen for incoming requests on port `3000`.**


---
## 5. Why Do We Need Node.js for Backend?

**Node.js** is a **JavaScript runtime** that allows JavaScript to run **outside the browser**, especially on a server.

Normally, JavaScript is commonly executed inside browsers like Chrome. For a JavaScript backend, we need to run JavaScript on the **server**, and **Node.js provides the runtime** to do that.

### 🔄 Basic Flow

**Client/Browser → HTTP Request → Node.js Backend → HTTP Response → Client/Browser**

### ⚙️ What Can Node.js Do?

Node.js allows backend JavaScript to:

- 📥 Receive HTTP requests
- 📤 Send JSON/HTML responses
- 🗄️ Connect to databases
- 📁 Read and write files
- 🔐 Authenticate users
- 🧠 Perform business logic

### 📁 Example: File System

Node.js provides built-in modules such as `fs` (**File System**) that allow JavaScript to work with files on the server.

### 💡 Important to Remember

**JavaScript** = Programming Language  
**Node.js** = Runtime that runs JavaScript outside the browser

> **Easy Definition:** Node.js lets us use JavaScript to build and run backend/server-side applications.

# JavaScript Functional Programming Basics

## 1. What is JavaScript?

**JavaScript** is a **multi-paradigm programming language** that supports **functional programming**.

> **Easy Definition:** JavaScript supports functional programming, where functions can be used and passed around like values.

---

## 2. What Does First-Class Citizen Mean?

In JavaScript, **functions are treated as first-class citizens**.

This means a function can be:

- Stored in a variable
- Passed as an argument to another function
- Returned from another function

### Example

`const greet = function() { return "Hello"; };`

> **Easy Definition:** Functions can be treated like other values in JavaScript.

---

## 3. What is a Pure Function?

A **pure function**:

- Gives the **same output for the same input**
- Does **not change anything outside the function** (no side effects)

### Example

`function add(a, b) { return a + b; }`

`add(2, 3)` → `5`  
`add(2, 3)` → `5`

It always gives the same result for the same input.

> **Easy Definition:** Same input → Same output, with no side effects.

---

## 4. What is a Higher-Order Function?

A **higher-order function** is a function that:

- Accepts another function as an argument, **and/or**
- Returns another function.

### Example

`function calculate(a, b, operation) {`  
`    return operation(a, b);`  
`}`

Here, `calculate()` is a **higher-order function** because it accepts another function through `operation`.

### 🔄 Callback Function

A **callback function** is a function that is **passed to another function so it can be called when needed**.

`function add(a, b) {`  
`    return a + b;`  
`}`

`function subtract(a, b) {`  
`    return a - b;`  
`}`

`function multiply(a, b) {`  
`    return a * b;`  
`}`

`function calculate(a, b, operation) {`  
`    return operation(a, b);`  
`}`

`calculate(10, 5, add);       // 15`  
`calculate(10, 5, subtract);  // 5`  
`calculate(10, 5, multiply);  // 50`

### 💡 Relationship

`calculate()` → **Higher-Order Function**  
`add` → **Callback Function**  
`subtract` → **Callback Function**  
`multiply` → **Callback Function**  
`operation` → Parameter that receives the callback

> **Easy Rule:**  
> Higher-Order Function = **Receives the function**  
> Callback Function = **The function being passed**

---

## 5. What is a Closure?

A **closure** happens when an inner function **remembers and can access variables from its outer function**, even after the outer function has finished executing.

### Example

`function outerFunction() {`  
`    let age = 90;`

`    function innerFunction() {`  
`        console.log(age);`  
`    }`

`    return innerFunction;`  
`}`

`const result = outerFunction();`

`result(); // 90`


### How It Works

`outerFunction()` creates:

`age = 90`

Then it returns:

`innerFunction`

So:

`const result = outerFunction();`

means `result` now holds the returned `innerFunction`.

Later:

`result();`

runs `innerFunction()`, and it still remembers:

`age = 90`

#Express Js and Server

## 1. What is Express.js?

**Express.js** is a **web framework for Node.js** that makes it easier to build **web servers and APIs**.

---

## 2. What is a Query Parameter?

A **query parameter** is extra information sent by the client **after `?` in the URL**, often used for filtering, searching, or sorting.

### Example

`http://localhost:3000/student?id=10`

Here:

- `/student` → Route
- `id` → Query parameter
- `10` → Value

In Express:

`req.query.id`

returns:

`10`

---

## 3. What is a Route Parameter?

A **route parameter** is a dynamic value that is **part of the URL path**, often used to identify a specific resource.

### Example

`http://localhost:3000/student1/123`

The Express route can be:

`/student1/:id`

Here:

- `:id` → Placeholder
- `123` → Value received from the client

In Express:

`req.params.id`

returns:

`123`

---

## 4. Query Parameter vs Route Parameter

**Query Parameter:**

`/student?id=10`

Access with:

`req.query.id`

**Route Parameter:**

`/student1/10`

Access with:

`req.params.id`

> **Remember:**  
> Query → `req.query`  
> Route → `req.params`

---

## 5. Simple Express Server with Query and Route Parameter

`const express = require('express');`

`const app = express();`

`const data = {`  
`    name: 'sayem emon',`  
`    age: 29,`  
`    district: 'Noakhali'`  
`};`

### GET Request

`app.get('/home', function(req, res) {`  
`    res.send('This is Sayem home page');`  
`});`

When the client sends:

`GET /home`

The server responds:

`This is Sayem home page`

---

### Query Parameter

`app.get('/student', function(req, res) {`  
`    console.log(req.query.id);`  
`    res.json(data);`  
`});`

Client requests:

`http://localhost:3000/student?id=10`

Then:

`req.query.id` → `10`

---

### Route Parameter

`app.get('/student1/:id', function(req, res) {`  
`    console.log(req.params.id);`  
`    res.send('Student ID is ' + req.params.id);`  
`});`

Client requests:

`http://localhost:3000/student1/10`

Then:

`req.params.id` → `10`

Server responds:

`Student ID is 10`

---

### Start the Server

`app.listen(3000);`

This starts the Express server and makes it **listen for requests on port `3000`**.

---

## 💡 Important to Remember

**GET** → Client wants to retrieve something from the server.

**`req` (Request)** → Information coming **from the client**

**`res` (Response)** → Information going **back to the client**

**`req.query`** → Access query parameters

**`req.params`** → Access route parameters

**`res.send()`** → Send a response

**`res.json()`** → Send JSON data

**`app.listen(3000)`** → Start listening for requests on port `3000`

# CommonJS vs ES Modules — Require & Import

## 1. What is the Difference Between CommonJS and ES Modules?

### CommonJS (CJS)

- Uses `require()` to import
- Uses `module.exports` to export
- `require()` loads modules **synchronously**
- Traditional module system used by Node.js

### ES Modules (ESM)

- Uses `import` to import
- Uses `export` to export
- Modern **standard JavaScript module system**
- Supports static analysis and modern module loading

> **Remember:**  
> CommonJS → `require()` + `module.exports`  
> ES Modules → `import` + `export`

---

## 2. CommonJS — Export & Require

### Export

`math_require.js`

`function add(a, b) {`  
`    return a + b;`  
`}`

`module.exports = add;`

### Require

`app_require.js`

`const add = require('./math_require');`

`console.log(add(10, 5));`

Output → `15`

> `module.exports` → Makes code available  
> `require()` → Brings that code into another file

---

## 3. ES Module — Export & Import

### Export

`subtract_import.js`

`function sub(a, b) {`  
`    return a - b;`  
`}`

`export default sub;`

### Import

`app_import.js`

`import sub from './subtract_import.js';`

`console.log(sub(10, 5));`

Output → `5`

> `export` → Makes code available  
> `import` → Brings that code into another file

---

## 4. Can We Mix CommonJS and ES Modules?

**Yes, Node.js provides interoperability, but mixing them can require special handling.**

For simple projects, it is better to use **one module system consistently**.

---

## 5. Advantages of ES Modules Over CommonJS

- **Static Analysis** → Imports/exports can be analyzed before execution
- **Standardization** → Official JavaScript module standard
- **Better Tooling** → Supports features such as tree shaking
- **Browser Support** → Works natively in modern browsers
- **Modern Module Loading** → Better suited to modern JavaScript environments

---

## 6. Why is `require()` Synchronous?

`require()` loads the module and waits for it to finish before continuing.

**Flow:**

`require()` → Load Module → Return Exports → Continue

ES Modules declare dependencies ahead of execution, allowing the runtime to load and link modules more flexibly.

---

## 💡 Quick Interview Summary

**CommonJS:**  
`module.exports` → Export  
`require()` → Import

**ES Modules:**  
`export` → Export  
`import` → Import

**Easy Definition:** Both `require()` and `import` bring code from another module/file into the current file, but they belong to different module systems.

# What is a Thread?

A **thread** is the **smallest unit of execution inside a process** that performs tasks.

> **Process = A program that is currently running**  
> **Thread = A unit of execution/work inside that process**

Think of a restaurant:

**Process = Restaurant**  
**Thread = Worker**

#What is Event Loop?



> **Easy Definition:** The Event Loop coordinates ready asynchronous tasks and lets their callbacks run when the JavaScript thread is available.

JavaScript executes code on the **main thread one task at a time**.

When JavaScript gets an **asynchronous task**, it does not make the main thread wait. The async work is **offloaded/handled outside the current JavaScript execution** by Node.js, browser APIs, or the OS.

When the async work is finished, its **callback/task becomes ready in a queue**.

The **Event Loop** checks when the main JavaScript thread can handle more work and allows the ready callback to execute.

### 🔄 Simple Flow

**Main Thread → Async Work → Offload/Handle Outside → Queue → Event Loop → Main Thread**



### 💡 Remember

- **Main Thread** → Executes JavaScript one task at a time
- **Async Work** → Can be handled/offloaded outside the current JS execution
- **Queue** → Holds ready callbacks/tasks
- **Event Loop** → Coordinates when ready callbacks can execute
- **Event Loop does NOT perform the heavy work itself**

# Promise, Async & **Await**

## 1. What is a Promise?

A **Promise** is an object that represents the **future result of an asynchronous operation**.

A Promise has 3 states:

- **Pending** → Still waiting
- **Fulfilled** → Success → `resolve()`
- **Rejected** → Failed → `reject()`

> **Easy Definition:** Promise = A result that will come later, either success or failure.

---

## 2. What is `async`?

`async` is used to define an **asynchronous function**.

- An `async` function **always returns a Promise**.
- It allows us to use `await` inside the function.

> **Easy Definition:** `async` = Defines an async function that returns a Promise and can use `await`.

---

## 3. What is `await`?

`await` is used inside an `async` function to **wait for a Promise to settle**.

If the Promise is pending:

1. `await` pauses that **async function**.
2. The main thread can continue other work.
3. When the Promise settles, `await` gets the result.
4. The async function continues.

> **Easy Definition:** `await` = Pause this async function → Wait for Promise → Get result → Continue.

---

# 🍕 Easy Pizza Example

`// Returns a Promise`

`function orderPizza() {`

&nbsp;&nbsp;&nbsp;&nbsp;`return new Promise((resolve, reject) => {`

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;`setTimeout(() => {`

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;`}, 3000);`

&nbsp;&nbsp;&nbsp;&nbsp;`});`

`}`


`// Async function`

`async function eatPizza() {`

&nbsp;&nbsp;&nbsp;&nbsp;`console.log("Ordering pizza...");`

&nbsp;&nbsp;&nbsp;&nbsp;`const pizza = await orderPizza();`

&nbsp;&nbsp;&nbsp;&nbsp;`console.log(pizza);`

&nbsp;&nbsp;&nbsp;&nbsp;`console.log("Eating pizza!");`

`}`


`eatPizza();`

`console.log("I can use my phone while waiting");`

### Output

`Ordering pizza...`  
`I can use my phone while waiting`

**After 3 seconds:**

`Pizza is ready!`  
`Eating pizza!`

---

## 🔄 How `await` Works Here

**`orderPizza()` is called**

↓  

**Returns a Promise**

↓  

**Promise is Pending**

↓  

**`await` pauses `eatPizza()`**

↓  

**Main thread can do other work**

↓  

`I can use my phone while waiting`

↓  

**Pizza becomes ready**

↓  

`resolve("Pizza is ready!")`

↓  

**Promise = Fulfilled**

↓  

**`await` gets the result**

↓  

`pizza = "Pizza is ready!"`

↓  

**`eatPizza()` continues**

**Promise** → Result will come later

**`async`** → Function returns a Promise and can use `await`

**`await`** → Pause async function → Wait for Promise → Get result → Continue

> **Important:** `await` pauses the async function, **not the entire JavaScript main thread**.
